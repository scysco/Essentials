# 🌐 Proyecto: Red de Optimización SoftWorks (MST)

Este notebook implementa la solución para el caso de estudio de **SoftWorks S.A.**.

Se abordan dos problemas:
1.  **Estructura Organizacional:** Modelado mediante un Grafo Dirigido.
2.  **Infraestructura de Red:** Optimización de costos mediante el algoritmo de Kruskal (Árbol de Expansión Mínima).

```python
import networkx as nx
import matplotlib.pyplot as plt

# Configuración para mostrar gráficos en el notebook
%matplotlib inline
```

## 1. Organigrama Jerárquico

Visualización de la estructura de mando de la empresa, desde la Dirección General (DG) hasta las áreas operativas.

```python
def generar_organigrama():
    G = nx.DiGraph() # Grafo Dirigido (Con flechas)

    # Definición exacta según Caso de Estudio
    # Nivel 1
    G.add_edge("DG", "DES")
    G.add_edge("DG", "ADM")
    # Nivel 2 (Rama Desarrollo)
    G.add_edge("DES", "FE")
    G.add_edge("DES", "BE")
    # Nivel 2 (Rama Admin)
    G.add_edge("ADM", "FIN")
    # Nivel 3 (Rama Front-end)
    G.add_edge("FE", "UI")
    G.add_edge("FE", "QA")
    # Nivel 3 (Rama Back-end)
    G.add_edge("BE", "API")
    G.add_edge("BE", "DB")

    # Posición Manual para jerarquía visual
    pos = {
        'DG': (0, 10),
        'DES': (-4, 7), 'ADM': (4, 7),
        'FE': (-6, 4), 'BE': (-2, 4), 'FIN': (4, 4),
        'UI': (-7, 1), 'QA': (-5, 1), 'API': (-3, 1), 'DB': (-1, 1)
    }

    plt.figure(figsize=(10, 7))
    plt.title("Estructura Jerárquica SoftWorks S.A.", fontsize=12, fontweight='bold')
    
    # Dibujar el grafo
    nx.draw(G, pos, 
            with_labels=True, 
            node_size=3000, 
            node_color='lightblue', 
            font_size=9, 
            font_weight='bold', 
            arrows=True, 
            arrowsize=20)
    
    plt.axis('off')
    plt.show()

generar_organigrama()
```

## 2. Red de Servidores (Optimización de Costos)

Se busca conectar 6 servidores minimizando el costo total del cableado de fibra óptica. Se utiliza un grafo ponderado no dirigido.

```python
def generar_redes_servidores():
    G = nx.Graph() # Grafo NO Dirigido

    # Nodos (Servidores)
    servidores = ['S1', 'S2', 'S3', 'S4', 'S5', 'S6']
    G.add_nodes_from(servidores)

    # Aristas con Costos (en miles de pesos)
    conexiones = [
        ('S1', 'S2', 4), ('S1', 'S3', 6),
        ('S2', 'S3', 5), ('S2', 'S4', 7),
        ('S3', 'S4', 8), ('S3', 'S5', 3),
        ('S4', 'S5', 2), ('S4', 'S6', 9),
        ('S5', 'S6', 1)
    ]
    G.add_weighted_edges_from(conexiones)

    # Posición Geométrica Simétrica
    pos = {
        'S1': (0, 5),   # DG (Arriba)
        'S2': (-4, 2),  # DES
        'S3': (4, 2),   # FE
        'S4': (-4, -2), # BE
        'S5': (4, -2),  # UI
        'S6': (0, -5)   # FIN (Abajo)
    }

    # Calcular MST (Algoritmo Kruskal)
    mst = nx.minimum_spanning_tree(G, algorithm='kruskal', weight='weight')
    costo_total = mst.size(weight='weight')
    
    print(f"Costo Total Mínimo de la Red: ${costo_total} (miles de pesos)")

    # --- Visualización ---
    plt.figure(figsize=(10, 8))
    plt.title(f"Red Óptima (MST) - Costo Total: {costo_total}", fontsize=12, fontweight='bold')

    # 1. Dibujar todas las conexiones posibles (fondo gris claro)
    nx.draw_networkx_edges(G, pos, edge_color='lightgray', style='dashed', width=1.5)

    # 2. Dibujar las conexiones del MST (resaltadas)
    nx.draw_networkx_edges(mst, pos, edge_color='#44546A', width=4)

    # 3. Dibujar nodos
    nx.draw_networkx_nodes(G, pos, node_size=1300, node_color='#FFA500', edgecolors='black')
    nx.draw_networkx_labels(G, pos, font_weight='bold', font_size=11)

    # 4. Etiquetas de los costos
    labels = nx.get_edge_attributes(G, 'weight')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=labels, font_size=12, font_color='blue')

    plt.axis('off')
    plt.show()

generar_redes_servidores()
```
