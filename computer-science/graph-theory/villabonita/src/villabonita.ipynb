# 🚑 Proyecto: Red de Emergencia "Villabonita" (Dijkstra)

Este notebook resuelve el problema de logística de emergencia para la ciudad de **Villabonita del Grijalva**.

**Objetivo:** Determinar las rutas más rápidas desde el Centro de Acopio (A) hacia los refugios, utilizando un Grafo Dirigido Ponderado (tiempos en minutos) y el algoritmo de **Dijkstra**.

```python
import networkx as nx
import matplotlib.pyplot as plt

# Configuración visual
%matplotlib inline
```

## 1. Definición de la Red Vial

Se crea un Grafo Dirigido (`DiGraph`) donde los pesos representan los minutos de viaje según el reporte de reconocimiento.

```python
def generar_red_emergencia():
    G = nx.DiGraph()

    # Definición de Aristas (Origen, Destino, Peso/Minutos)
    rutas = [
        ('A', 'B', 20), ('A', 'C', 15), ('A', 'D', 45),
        ('B', 'C', 10), ('B', 'E', 30),
        ('C', 'D', 10), ('C', 'E', 35), ('C', 'F', 12), ('C', 'A', 18),
        ('D', 'A', 40),
        ('E', 'F', 5),  ('E', 'G', 20),
        ('F', 'G', 15), ('F', 'B', 25),
        ('G', 'A', 30)
    ]
    
    G.add_weighted_edges_from(rutas)
    return G

G = generar_red_emergencia()
```

## 2. Cálculo de Rutas Más Rápidas (Dijkstra)

Calculamos el tiempo mínimo y el camino exacto desde el Nodo **A** hacia todos los demás nodos.

```python
origen = 'A'
destinos = ['B', 'C', 'D', 'E', 'F', 'G']

print(f"--- Tiempos mínimos desde Centro de Acopio ({origen}) ---")

for destino in destinos:
    try:
        tiempo = nx.dijkstra_path_length(G, origen, destino, weight='weight')
        ruta = nx.dijkstra_path(G, origen, destino, weight='weight')
        print(f"A -> {destino}: {tiempo} minutos | Ruta: {' -> '.join(ruta)}")
    except nx.NetworkXNoPath:
        print(f"A -> {destino}: No hay ruta disponible.")
```

## 3. Visualización de la Red

Generamos un gráfico para visualizar la topología de la red y las direcciones de flujo.

```python
def dibujar_grafo(G):
    pos = nx.circular_layout(G)  # Layout circular para ver bien las conexiones
    
    plt.figure(figsize=(10, 8))
    plt.title("Red Logística de Emergencia: Villabonita del Grijalva", fontsize=14, fontweight='bold')
    
    # Dibujar Aristas
    nx.draw_networkx_edges(G, pos, arrowstyle='-|>', arrowsize=20, edge_color='gray', width=1.5, connectionstyle='arc3, rad=0.1')
    
    # Dibujar Nodos
    nx.draw_networkx_nodes(G, pos, node_size=800, node_color='#FF6347', edgecolors='black')
    nx.draw_networkx_labels(G, pos, font_size=12, font_weight='bold', font_color='white')
    
    # Etiquetas de pesos (tiempos)
    labels = nx.get_edge_attributes(G, 'weight')
    # Ajuste manual simple para etiquetas en grafos dirigidos (puede solaparse en arc3)
    nx.draw_networkx_edge_labels(G, pos, edge_labels=labels, font_size=10, label_pos=0.3)
    
    plt.axis('off')
    plt.show()

dibujar_grafo(G)
```
